```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt

urls = ["https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRvurgcqDycsu4z09jcZONjSioLsAL6SgZdy4gM9Mw9BIZarf-5VuOJnq4&s=10"]

for url in urls:

  image = io.imread(url)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

height = image_2.shape[0]
width = image_2.shape[1]

thick = 1.0

m1 = (height - 1) / (width - 1)
m2 = -(height - 1) / (width - 1)

for y in range(height):
    for x in range(width):
      y1 = m1 * x
      selisih1 = abs(y - y1)

      y2 = m2 * x + (height - 1)
      selisih2 = abs(y - y2)

      if selisih1 <= thick or selisih2 <= thick:

        image_3[y , x] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt

urls = ["https://i.pinimg.com/736x/26/ef/12/26ef127ba199f0c5f0ab10fff6b1452e.jpg"]

for url in urls :

  image = io.imread(url)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

  tinggi = image_2.shape[0]
  print("Tingginya = ", tinggi)
  lebar = image_2.shape[1]
  print("Lebar = ", lebar)
  radius = 100
  print("Radius = ", radius)
  xcent, ycent = 250, 280

  for y in range (tinggi):
    for x in range (lebar):
      dx = abs(x - xcent)
      dy = abs(y - ycent)

      if dx <= radius and dy <= radius:
        image_3[y,x] = [255,255,255]

# y --> baris
# x --> kolom

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
